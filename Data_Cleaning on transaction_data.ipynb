```python
import numpy as np
import pandas as pd
```

```python
b = pd.read_csv('/content/transaction_data.csv',encoding='latin-1')
b.head()
```

```python
# Step 1: Renaming the columns
a={'Transaction ID':'Transac_id'}
b.rename(columns=a,inplace=True)
b
```

```python
# STEP 2 : REMOVING DUPLICATE VALUES
b.drop_duplicates(inplace=True)
b
```

```python
# STEP 3: Changing the data present in a particular columns
b['Device Used'].unique()
```

```python
data={'Desktop':'Desk', 'Mobile':'Mob'}
b['Device Used'] = b['Device Used'].map(data)
b['Device Used'].unique()
b
```

```python
# STEP 4: Remove unwanted data present in a column

b['Network Slice ID'] = b['Network Slice ID'].str.strip('Slice')
b
```

```python
# STEP 5: Applying inbuilt functions on particular column
b['Device Used'] = b['Device Used'].str.upper()
b['Device Used'].unique()
```

```python
b.info()
```

```python
b['Device Used'] = b['Device Used'].str.lower()
b['Device Used'].unique()
```

```python
# STEP 6 : How to handle the missing values

c = b.dropna( how = 'all')
c
```

```python
# 6.1 Drop only if atleast one NaN exists in a row
c = b.dropna( how = 'any')  # If any one of the values in a row is NaN it will be dropped
c
```

```python
# 6.2) Filling NaN values with zero, mean or median

b.fillna(0)
b
```

```python
e = b['Transaction Amount'].mean()
print(e)
```

```python
b['Transaction Amount'] = b['Transaction Amount'].fillna(e)
b
```

```python
d = b['Transaction Amount'].median()

d
```

```python
b['Transaction Amount'] = b['Transaction Amount'].fillna(d)

b
```

```python
b.info()
```

```python
# STEP 7 : Changing the datatype of a column

b['Network Slice ID'] = b['Network Slice ID'].astype(int)
b.info()
```

```python
# STEP 8: Rearranging the order of columns

b.head()
```

```python
new_order = [1,3,4,2,0]

b = b[b.columns[new_order]]

b.head()
```

```python
# STEP 9 : Changing the data present in a particular column

b['Transaction Type'] = b['Transaction Type'].str.replace('Deposit','Deposits')

b
```

```python
# STEP 10: Dropping unwanted column
del b['Transaction Type']
b
```

```python
f = b.drop(0, axis = 0)

f  # It will remove the mentioned index positions (row)
```

```python
g = b.drop('Transac_id', axis = 1)

g  # It will remove the mentioned column
```

```python
# STEP 11 : Dropping unwanted rows
b.drop('Receiver Account ID', axis=1, inplace = True)
b.head()
```

```python
b.drop('Transac_id', axis=1, inplace = True)
b
```

```python
# STEP 12 : Resetting the index values
b.reset_index(inplace=True)
b
```

```python
# STEP 13: Setting the column name to index positions slot
b.set_index('index', inplace=True)
b
```
