```python
#ARRAY CREATION WAYS
# 1) Zeros ---> #it is used to fill folat based zeros to the given rows and columns
# Syntax : np.zeros((row,columns))
np.zeros((4,5))
```

```python
# 1)Ones ---> it is used to fill folat based ones to the given rows and columns
#syntax : np.ones(row,columns)
np.ones((4,5))
```

```python
# FULL ---> it is used to fill the given value
# Syntax : np.full(rows,column),value)
np.full((4,5),5)
```

```python
#3 way
#eye ---> it is used to get identity matrix from the identity matrix from the given values
#syntax : np.eye(n) where n is number rows and columns
np.eye(3)
```

```python
#Linspace : it is used to create the number of divisions between the given values
#syntax : np.linspace(start,stop,num)
np.linspace(0,1,7)
```

```python
#7 rand --> it is fill random float values in given rows and columns
#note but the values lies between 0 and 1
#syntax np.random.rand(row,col)
np.random.rand(2,3)
```

```python
# 5) randint ---> it is used to fill random values from the given range to given rows and columns
# syntax : np.randint(value1,value2, (row,columns))
np.random.randint(1,10, (2,3))
```

```python
# DATA FILTERING (DEEP COPY) --> The process of obtaining the new address by dropping the old address of an arrray
# MASKING (SHALLOW COPY) --> The process of new layering the main address of an array

a=np.array([1,2,3,4])
b=a.view()   #DEEP COPY : view is the operation used to perform a Deep copy
print(b)
```
