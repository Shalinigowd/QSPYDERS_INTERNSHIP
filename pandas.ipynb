```python
b = pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/amazon_fires.csv', encoding='latin-1')
b
```

```python
#DATA CLEANING

# STEP 1: Renaming the columns
# SYNTAX : var1 = {'old_col1':'new_col1',........,'old_colN':'new_colN'}
         # var.rename(columns = var1, inplace = True)

#inplace = True ---> It will permanantly change the data
```

```python
data ={
    'ano':'Year',
    'mes':'Month',
    'estado':'State',
    'numero':'No_of_fires',
    'encontro':'Date'
}
b.rename(columns=data,inplace=True)
b
```

```python
# STEP 2: REMOVING DUPLICATE VALUES
#SYNTAX: var.drop_duplicates(inplace=True)


c = pd.DataFrame({
    'name': ['sushanth','jaya','jaya','moni','nagaraj'],
    'roll_no' : [80,46,46,82,67],
    'marks' : [100,80,80,86,90],
    'gender' : ['M','F','F','M','M']
})
c
```

```python
c.drop_duplicates(inplace=True)
c
#if bothe the rows are having same functionalities , then it will be dropped
```

```python
# STEP 3: CHANGING THE DATA PRESENT IN A PARTICULAR COLUMN
# SYNTAX : var['col_name']= var['col_name'].map(dict)
b['Month'].unique()   #extractig unique values from a column
```

```python
data = {'Janeiro':'January', 'Fevereiro':'February', 'Março':'March', 'Abril':'April', 'Maio':'May', 'Junho':'June', 'Julho':'July',
       'Agosto':'August', 'Setembro':'September', 'Outubro':'October', 'Novembro':'November', 'Dezembro':'December'}

b['Month'] = b['Month'].map(data)
b['Month'].unique()
```

```python
# STEP 4: REMOVE UNWANTED DATA PRESENT IN A COLUMN
#SYNTAX: var['col_name']=var['col_name'].str.strip('Unwanted data')
b['No_of_fires']=b['No_of_fires'].str.strip('Fires')
b
```

```python
# STEP 5: Applying inbuilt functions on particular column

#Syntax: var['col_name']=var['col_name'].str.fname()

b['Month']=b['Month'].str.upper()
b['Month'].unique()
```
